```python
def any_lowercase2(s):
    for c in s:
        if 'c'.islower():
            return 'True'
        else:
            return 'False' #Always returns true - checking whether string 'c' is lowercase
```

```python
any_lowercase2("Trend")
```

```python
any_lowercase2("banana")
```

```python
def any_lowercase3(s):
    for c in s:
        flag = c.islower()
    return flag #Loops through every character but overwrites every time - only reporting on last character
```

```python
any_lowercase3("Banana")
```

```python
any_lowercase3("banana")
```

```python
def any_lowercase4(s):
    flag = False
    for c in s:
        flag = flag or c.islower()
    return flag   #Looking for ANY lowercase letters.
```

```python
any_lowercase4("Banana")
```

```python
any_lowercase4("banana")
```

```python
any_lowercase4("BANANA")
```

```python
def any_lowercase5(s):
    for c in s:
        if not c.islower():
            return False
    return True  #Looks for any uppercase letters
```

```python
def rotate_letter(letter, n):
    
    if letter.isupper():
        start = ord('A')
    elif letter.islower():
        start = ord('a')
    else:
        return letter

    c = ord(letter) - start
    i = (c + n) % 26 + start
    return chr(i)


def rotate_word(word, n):
    res = ''
    for letter in word:
        res += rotate_letter(letter, n)
    return res


    
```

```python
rotate_word("sleep", 10)
```
