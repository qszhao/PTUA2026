```python
import random
import math
import matplotlib.pyplot as plt
```

```python
pi = math.pi
```

```python
convergence = 0.01    

pi_estimates1 = []

for i in range(10):
    n = 0 # number of points falling in the unit circle
    d = 0 # number of points falling in the unit square
    simulating = True # use as a sentinel

    while simulating:
        x = random.random()
        y = random.random()
        if x**2 + y**2 <= 1.0:
            n += 1
        d += 1
        ratio = 4 * n * 1./d
        percent_diff = abs(ratio - math.pi) / math.pi
    
        if percent_diff <= convergence:
            pi_estimates1.append((ratio, d))
            break

    print(f"0.01 Simulation {i+1}: {ratio} after {d} draws")
    
```

```python
draw_amounts = [draw_amount1, draw_amount2, draw_amount3, draw_amount4]
convergence = [0.01, 0.001, 0.0001, 0.00001]

for conv, draws in zip(convergence, draw_amounts):
    n = len(draws)
    average = sum(draws) / n
    std_dev = (sum((x - average) ** 2 for x in draws) / (n - 1)) ** 0.5
    print(f"Convergence: {conv}, Average: {average}, Standard Deviation: {std_dev}")

```
